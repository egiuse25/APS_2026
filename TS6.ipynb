Posteriormente se hace la verificación de la respuesta en frecuencia en módulo y fase junto con el diagrama de polos y ceros de todos los sistemas con Numpy.

```python
'EJERCICIO A'
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

b = [1, 1, 1, 1] 
a = [1, 0, 0, 0] 

w, h = signal.freqz(b, a, worN=511) 

modulo = np.abs(h)

fase = np.unwrap(2 * np.angle(h)) / 2 

ceros, polos, ganancia = signal.tf2zpk(b, a)


fig = plt.figure(figsize=(12, 8))

ax1 = plt.subplot(2, 2, 1)
ax1.plot(w, modulo, 'b') 
ax1.set_title('Módulo del Ejercicio a)')
ax1.set_ylabel('Amplitud')
ax1.set_xlabel('Frecuencia')
ax1.grid(True)

ax2 = plt.subplot(2, 2, 3)
ax2.plot(w, fase, 'g')
ax2.set_title('Fase del Ejercicio a)')
ax2.set_ylabel('Fase [rad]')
ax2.set_xlabel('Frecuencia')
ax2.grid(True)

ax3 = plt.subplot(1, 2, 2)
circulo = plt.Circle((0, 0), 1, color='black', fill=False, linestyle='--')
ax3.add_patch(circulo)

ax3.scatter(np.real(ceros), np.imag(ceros), s=100, marker='o', facecolors='none', edgecolors='b', label='Ceros')
ax3.scatter(np.real(polos), np.imag(polos), s=100, marker='x', color='r', label='Polos')

ax3.axhline(0, color='black', lw=1)
ax3.axvline(0, color='black', lw=1)
ax3.set_xlim([-1.5, 1.5])
ax3.set_ylim([-1.5, 1.5])
ax3.set_aspect('equal')
ax3.set_title('Diagrama de Polos y Ceros del Ejercicio a)')
ax3.set_xlabel('Eje Real')
ax3.set_ylabel('Eje Imaginario')
ax3.legend()
ax3.grid(True)

plt.tight_layout()
plt.show()


'EJERCICIO B'
b = [1, 1, 1, 1, 1] 
a = [1, 0, 0, 0, 0] 

w, h = signal.freqz(b, a, worN=512) 

modulo = np.abs(h)

fase = np.unwrap(2 * np.angle(h)) / 2 

ceros, polos, ganancia = signal.tf2zpk(b, a)

fig = plt.figure(figsize=(12, 8))

ax1 = plt.subplot(2, 2, 1)
ax1.plot(w, modulo, 'b') 

ax1.set_title('Módulo del Ejercicio b)')
ax1.set_ylabel('Amplitud')
ax1.set_xlabel('Frecuencia')
ax1.grid(True)

ax2 = plt.subplot(2, 2, 3)
ax2.plot(w, fase, 'g')
ax2.set_title('Fase del Ejercicio b)')
ax2.set_ylabel('Fase [rad]')
ax2.set_xlabel('Frecuencia')
ax2.grid(True)

ax3 = plt.subplot(1, 2, 2)
circulo = plt.Circle((0, 0), 1, color='black', fill=False, linestyle='--')
ax3.add_patch(circulo)

ax3.scatter(np.real(ceros), np.imag(ceros), s=100, marker='o', facecolors='none', edgecolors='b', label='Ceros')
ax3.scatter(np.real(polos), np.imag(polos), s=100, marker='x', color='r', label='Polos')

ax3.axhline(0, color='black', lw=1)
ax3.axvline(0, color='black', lw=1)
ax3.set_xlim([-1.5, 1.5])
ax3.set_ylim([-1.5, 1.5])
ax3.set_aspect('equal')
ax3.set_title('Diagrama de Polos y Ceros del Ejercicio b)')
ax3.set_xlabel('Eje Real')
ax3.set_ylabel('Eje Imaginario')
ax3.legend()
ax3.grid(True)

plt.tight_layout()
plt.show()

'EJERCICIO C'
b = [1, -1] 
a = [1, 0, 0, 0, 0] 

w_custom = np.linspace(1e-5, np.pi, 512)

w, h = signal.freqz(b, a, worN=w_custom)

modulo = np.abs(h)

fase = np.unwrap(2 * np.angle(h)) / 2 

ceros, polos, ganancia = signal.tf2zpk(b, a)

fig = plt.figure(figsize=(12, 8))

ax1 = plt.subplot(2, 2, 1)
ax1.plot(w, modulo, 'b') 
ax1.set_title('Módulo del Ejercicio c)')
ax1.set_ylabel('Amplitud')
ax1.set_xlabel('Frecuencia')
ax1.grid(True)

ax2 = plt.subplot(2, 2, 3)
ax2.plot(w, fase, 'g')
ax2.set_title('Fase del Ejercicio c)')
ax2.set_ylabel('Fase [rad]')
ax2.set_xlabel('Frecuencia')
ax2.grid(True)

ax3 = plt.subplot(1, 2, 2)
circulo = plt.Circle((0, 0), 1, color='black', fill=False, linestyle='--')
ax3.add_patch(circulo)

ax3.scatter(np.real(ceros), np.imag(ceros), s=100, marker='o', facecolors='none', edgecolors='b', label='Ceros')
ax3.scatter(np.real(polos), np.imag(polos), s=100, marker='x', color='r', label='Polos')

ax3.axhline(0, color='black', lw=1)
ax3.axvline(0, color='black', lw=1)
ax3.set_xlim([-1.5, 1.5])
ax3.set_ylim([-1.5, 1.5])
ax3.set_aspect('equal')
ax3.set_title('Diagrama de Polos y Ceros del Ejercicio c)')
ax3.set_xlabel('Eje Real')
ax3.set_ylabel('Eje Imaginario')
ax3.legend()
ax3.grid(True)

plt.tight_layout()
plt.show()

'EJERCICIO D'
b = [1, 0, -1] 
a = [1, 0, 0] 

w_custom = np.linspace(1e-5, np.pi, 512)

w, h = signal.freqz(b, a, worN=w_custom)

modulo = np.abs(h)

fase = np.unwrap(2 * np.angle(h)) / 2 

ceros, polos, ganancia = signal.tf2zpk(b, a)

fig = plt.figure(figsize=(12, 8))

ax1 = plt.subplot(2, 2, 1)
ax1.plot(w, modulo, 'b') 
ax1.set_title('Módulo del Ejercicio d)')
ax1.set_ylabel('Amplitud')
ax1.set_xlabel('Frecuencia')
ax1.grid(True)

ax2 = plt.subplot(2, 2, 3)
ax2.plot(w, fase, 'g')
ax2.set_title('Fase del Ejercicio d)')
ax2.set_ylabel('Fase [rad]')
ax2.set_xlabel('Frecuencia')
ax2.grid(True)

ax3 = plt.subplot(1, 2, 2)
circulo = plt.Circle((0, 0), 1, color='black', fill=False, linestyle='--')
ax3.add_patch(circulo)

ax3.scatter(np.real(ceros), np.imag(ceros), s=100, marker='o', facecolors='none', edgecolors='b', label='Ceros')
ax3.scatter(np.real(polos), np.imag(polos), s=100, marker='x', color='r', label='Polos')

ax3.axhline(0, color='black', lw=1)
ax3.axvline(0, color='black', lw=1)
ax3.set_xlim([-1.5, 1.5])
ax3.set_ylim([-1.5, 1.5])
ax3.set_aspect('equal')
ax3.set_title('Diagrama de Polos y Ceros del Ejercicio d)')
ax3.set_xlabel('Eje Real')
ax3.set_ylabel('Eje Imaginario')
ax3.legend()
ax3.grid(True)

plt.tight_layout()
plt.show()
```

Los resultados hechos en papel y lápiz son coincidentes con los gráficos hechos en Python. Todos los incisos constan de filtros FIR debido a que tienen sus polos en el centro. La diferencia de cada uno de los ejercicios está en el tipo de simetría y cantidad de muestras que tiene cada uno. El primero tiene simetría par con un número de muestras par, esto es un FIR de tipo II. El segundo inciso también consta de una simetría par pero con una cantidad de muestras impar, esto es un FIR de tipo I. Por otro lado, tanto el tercer inciso como el cuarto contienen una simetría impar con una cantidad de muestras par para el primero e impar para el segundo, siendo entonces el tercer inciso un FIR de tipo IV y el cuarto de tipo III. Es debido a la antisimetría que el módulo está constituido por senos y hay un desfasaje de $\frac{π}{2}$ radianes en sus fases.
